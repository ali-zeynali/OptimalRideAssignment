```python
import pandas as pd

# Load your CSV file
df = pd.read_csv("Public_Passenger_Vehicle_Licenses_20251125.csv")
```

```python
# Drop rows where any of the three columns is missing or blank
filtered = df.dropna(subset=["Vehicle Make", "Vehicle Model", "Vehicle Model Year"])
filtered = filtered[
    (filtered["Vehicle Make"].astype(str).str.strip() != "") &
    (filtered["Vehicle Model"].astype(str).str.strip() != "") &
    (filtered["Vehicle Model Year"].astype(str).str.strip() != "")
]

# Keep only relevant columns
filtered = filtered[["Vehicle Make", "Vehicle Model", "Vehicle Model Year"]]

# Remove duplicate rows
unique_df = filtered.drop_duplicates()

print(len(unique_df))

# Save to a new CSV file
unique_df.to_csv("unique_make_model_year.csv", index=False)

unique_df.head()
```
