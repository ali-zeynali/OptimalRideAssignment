```python
# Import necessary libraries
import pandas as pd
import random
import numpy as np
```

```python
# Load the two CSV files into pandas DataFrames
rides_df = pd.read_csv('Taxi_Trips_(2024-)_20251125.csv')  # Replace with your actual path
cars_df = pd.read_csv('unique_make_model_year.csv')    # Replace with your actual path

```

```python
print(len(rides_df))
print(len(cars_df))
```

```python
# Ensure the cars_df is shuffled so the car selection is random
cars_df_shuffled = cars_df.sample(frac=1, random_state=42).reset_index(drop=True)

# Calculate CO2 emissions per km for each car (using the MPG column)
cars_df_shuffled['CO2 per km (kg)'] = (3.89 / cars_df_shuffled['MPG']) * 1609.34

# Generate random indices to assign to the rides dataset
random_car_indices = np.random.choice(cars_df_shuffled.index, size=len(rides_df), replace=True)

# Add car details to the rides dataset using the random indices
rides_df['Vehicle Make'] = cars_df_shuffled.loc[random_car_indices, 'Vehicle Make'].values
rides_df['Vehicle Model'] = cars_df_shuffled.loc[random_car_indices, 'Vehicle Model'].values
rides_df['Vehicle Model Year'] = cars_df_shuffled.loc[random_car_indices, 'Vehicle Model Year'].values
rides_df['MPG'] = cars_df_shuffled.loc[random_car_indices, 'MPG'].values
rides_df['CO2 emission (g/km)'] = cars_df_shuffled.loc[random_car_indices, 'CO2 per km (kg)'].values
# Save the resulting merged dataset to a new CSV file
rides_df.to_csv('texas_Oct_2024_merged.csv', index=False)

print("Merging completed and saved!")
```
