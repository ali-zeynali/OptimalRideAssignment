```python
# Import necessary libraries
import pandas as pd
import random
```

```python
# Load the two CSV files into pandas DataFrames
rides_df = pd.read_csv('Taxi_Trips_(2024-)_20251125.csv')  # Replace with your actual path
cars_df = pd.read_csv('unique_make_model_year.csv')    # Replace with your actual path

```

```python
# Function to randomly assign car details to each ride request and calculate CO2 emissions
def assign_random_car_and_calculate_co2(row):
    # Randomly select a car from the cars dataset
    random_car = cars_df.sample(n=1).iloc[0]  # Randomly selects one row
    # Extract car details
    vehicle_make = random_car['Vehicle Make']
    vehicle_model = random_car['Vehicle Model']
    vehicle_year = random_car['Vehicle Model Year']
    mpg = random_car['MPG']
    
    # Calculate CO2 emitted per kilometer (using the formula derived above)
    co2_per_km = (3.9 / mpg) * 1609.34  # CO2 in g per km

    # Add these values to the current row
    row['Vehicle Make'] = vehicle_make
    row['Vehicle Model'] = vehicle_model
    row['Vehicle Model Year'] = vehicle_year
    row['MPG'] = mpg
    row['CO2 emission (g/km)'] = co2_per_km
    
    return row

# Apply the function to each row in the rides DataFrame
rides_df = rides_df.apply(assign_random_car_and_calculate_co2, axis=1)

# Save the resulting merged dataset to a new CSV file
rides_df.to_csv('texas_Oct_2024_merged.csv', index=False)

print("Merging completed and saved!")
```
